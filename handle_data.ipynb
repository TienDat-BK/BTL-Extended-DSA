```python
import pandas as pd

df = pd.read_csv('dataset/data_semantic.csv')
df.head()
df['plagiarism_type'].unique()
```

```python
df_demo_for_duplicate = pd.concat([df[df["plagiarism_type"] == "verbatim"][:0],
                                   df[df["plagiarism_type"] == "summary"][:25],
                                   df[df["plagiarism_type"] == "paraphrase"][:15],
                                   df[df["plagiarism_type"].isna()][:15]], ignore_index=True)
df_demo_for_duplicate = pd.concat([df_demo_for_duplicate['source_doc'],
                                   df_demo_for_duplicate['susp_doc']], ignore_index=True)
df_demo_for_duplicate = df_demo_for_duplicate.to_frame(name='text')
df_demo_for_duplicate.to_csv('dataset/df_demo_for_duplicate.csv')

df_demo_for_duplicate
```

```python
df_ques = pd.read_csv('dataset/dataset_small.csv')
df_demo_for_senmatic = pd.concat([df_ques[df_ques['is_duplicate'] == 1][:30],
                                  df_ques[df_ques['is_duplicate'] == 0][:15]], ignore_index=True)
df_demo_for_senmatic = pd.concat([df_demo_for_senmatic['question1'],
                                  df_demo_for_senmatic['question2']])
df_demo_for_senmatic = df_demo_for_senmatic.to_frame(name='text')
df_demo_for_senmatic.to_csv('dataset/df_demo_for_senmatic.csv')
df_demo_for_senmatic
```
